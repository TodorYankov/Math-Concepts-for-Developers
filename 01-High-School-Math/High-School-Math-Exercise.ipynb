### (a) Linear function

$$
y = ax + b; \quad a,b \in \mathbb{R}
$$

This is the general form of a linear function.

- $x$ is the independent variable.
- $y$ is the dependent variable.
- $a$ is the slope of the line.
- $b$ is the y-intercept.
- $a,b \in \mathbb{R}$ means that $a$ and $b$ are real numbers.

### (b) Distance between two points

$$
d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}
$$

This formula calculates the Euclidean distance between two points
$(x_1, y_1)$ and $(x_2, y_2)$ in a two-dimensional coordinate system.

- $d$ is the distance between the two points.
- $(x_1, y_1)$ are the coordinates of the first point.
- $(x_2, y_2)$ are the coordinates of the second point.
- $x_2-x_1$ is the horizontal difference.
- $y_2-y_1$ is the vertical difference.

### (c) Arithmetic Mean

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n}x_i
$$

This formula calculates the arithmetic mean of a set of values.

- $\bar{x}$ is the arithmetic mean.
- $n$ is the number of values.
- $x_i$ is the value at position $i$.
- $\sum$ means that all values are added together.
- $i=1$ means that the summation starts from the first value.
- $n$ means that the summation ends with the last value.

### (d) Rotation Matrix

$$
R(\theta) =
\begin{pmatrix}
\cos\theta & -\sin\theta \\
\sin\theta & \cos\theta
\end{pmatrix}
$$

This is a two-dimensional rotation matrix.

- $R(\theta)$ is the rotation matrix.
- $\theta$ is the angle of rotation.
- $\sin\theta$ and $\cos\theta$ are the sine and cosine of the angle.
- The matrix can be used to rotate a point or vector in a two-dimensional coordinate system.

### (e) Definite Integral

$$
\int_a^b f(x)\,dx =
\lim_{n\to\infty}
\sum_{i=1}^{n} f(x_i)\Delta x
$$

This formula represents a definite integral as the limit of a sum.

- $a$ and $b$ are the lower and upper limits of integration.
- $f(x)$ is the function being integrated.
- $\Delta x$ represents the width of a small interval.
- $f(x_i)\Delta x$ represents the area of a small rectangle.
- $\sum$ adds the areas of the rectangles.
- As $n \to \infty$, the rectangles become increasingly narrow and the sum approaches the value of the definite integral.

### (f) Golden Ratio

$$
\phi =
1 +
\frac{1}{
1 + \frac{1}{
1 + \frac{1}{
1 + \cdots
}}}
$$

This is an infinite continued fraction representing the golden ratio.

- $\phi$ (phi) represents the golden ratio.
- The same fractional structure repeats infinitely.
- The golden ratio is approximately $1.618$.
- Because the structure repeats itself, it can also be written as:

$$
\phi = 1 + \frac{1}{\phi}
$$

Therefore:

$$
\phi^2 - \phi - 1 = 0
$$

and the positive solution is:

$$
\phi = \frac{1+\sqrt{5}}{2}
$$

### (g) Binomial Coefficient

$$
\binom{n}{k} = \frac{n!}{k!(n-k)!}
$$

This is the formula for the binomial coefficient.

- $n$ is the total number of elements.
- $k$ is the number of elements being selected.
- $\binom{n}{k}$ is read as "n choose k".
- $!$ denotes the factorial operation.
- The binomial coefficient represents the number of ways to choose
  $k$ elements from $n$ elements when the order does not matter.

### (h) Laplacian

$$
\nabla^2 f =
\frac{\partial^2 f}{\partial x^2}
+
\frac{\partial^2 f}{\partial y^2}
+
\frac{\partial^2 f}{\partial z^2}
$$

This formula represents the Laplacian of a function $f$ in three dimensions.

- $f$ is a function of the variables $x$, $y$, and $z$.
- $\partial$ denotes a partial derivative.
- $\frac{\partial^2 f}{\partial x^2}$ is the second partial derivative
  of $f$ with respect to $x$.
- The same applies to $y$ and $z$.
- $\nabla^2 f$ is called the Laplacian of $f$.
- It is the sum of the second partial derivatives with respect to
  the spatial variables.

### Problem 5. Intro to proofs: proof by induction
Look up what a _proof by mathematical induction_ is. I like [Brilliant](https://brilliant.org/wiki/writing-a-proof-by-induction/), but you can use whatever you like.

Now, look at this proposition (which we also call _theorem_).
We call **triangular numbers** numbers which represent objects arranged in a triangle, like this [example](https://byjus.com/maths/triangular-numbers/). We denote the $n$th triangular number by $T_n$.

We can see that one way to write $T_n$ is by summing the numbers from $1$ to $n$. Another way is:
$$ T_n = \frac{n(n+1)}{2}$$

Prove this by induction, starting with $n = 1$.

**Base case: $n = 1$**

For $n = 1$:

$$
T_1 = 1
$$

Using the formula:

$$
T_1 = \frac{1(1+1)}{2} = \frac{2}{2} = 1
$$

Therefore, the formula is true for $n = 1$.

**Induction hypothesis**

Assume that the formula is true for some positive integer $k$:

$$
T_k = 1 + 2 + \cdots + k = \frac{k(k+1)}{2}
$$

**Inductive step**

We need to prove that the formula is also true for $k+1$.

$$
T_{k+1} = T_k + (k+1)
$$

Using the induction hypothesis:

$$
T_{k+1}
=
\frac{k(k+1)}{2} + (k+1)
$$

Therefore:

$$
T_{k+1}
=
\frac{k(k+1) + 2(k+1)}{2}
$$

Factor out $(k+1)$:

$$
T_{k+1}
=
\frac{(k+1)(k+2)}{2}
$$

This is exactly the required formula for $n=k+1$.

Therefore, by mathematical induction,

$$
\boxed{T_n=\frac{n(n+1)}{2}}
$$

for every positive integer $n$.

````python
def intersection_over_union(box_a, box_b):
    x_left = max(box_a[0], box_b[0])
    y_top = max(box_a[1], box_b[1])
    x_right = min(box_a[2], box_b[2])
    y_bottom = min(box_a[3], box_b[3])

    intersection_width = max(0, x_right - x_left)
    intersection_height = max(0, y_bottom - y_top)
    intersection_area = intersection_width * intersection_height

    area_a = (box_a[2] - box_a[0]) * (box_a[3] - box_a[1])
    area_b = (box_b[2] - box_b[0]) * (box_b[3] - box_b[1])

    union_area = area_a + area_b - intersection_area

    return intersection_area / union_area

ground_truth = (1, 1, 4, 4)
prediction = (2, 2, 5, 5)

iou = intersection_over_union(ground_truth, prediction)

print("IoU:", iou)    
````

### Huffman Compression Algorithm

#### 1. Lossless vs. Lossy Compression

**Lossless compression** allows the original data to be reconstructed
exactly after decompression.

$$
D(C(x)) = x
$$

Huffman coding is a lossless compression algorithm.

**Lossy compression** removes some information in order to achieve
stronger compression. The reconstructed data is therefore only an
approximation of the original:

$$
D(C(x)) \approx x
$$

Lossy compression can be acceptable when exact reconstruction is not
required, for example for some images, audio, and video.

````python
text = "AAAAABBC"

frequencies = Counter(text)
total = len(text)

entropy = 0

for symbol, frequency in frequencies.items():
    probability = frequency / total
    entropy -= probability * math.log2(probability)

    print(
        symbol,
        "frequency =", frequency,
        "probability =", probability
    )

print("Entropy:", entropy, "bits/symbol")
````

````python
class HuffmanNode:
    def __init__(self, char=None, frequency=0):
        self.char = char
        self.frequency = frequency
        self.left = None
        self.right = None

    def __lt__(self, other):
        return self.frequency < other.frequency

def build_huffman_tree(text):
    frequencies = Counter(text)

    heap = [
        HuffmanNode(char, frequency)
        for char, frequency in frequencies.items()
    ]

    heapq.heapify(heap)

    while len(heap) > 1:
        left = heapq.heappop(heap)
        right = heapq.heappop(heap)

        parent = HuffmanNode(
            frequency=left.frequency + right.frequency
        )

        parent.left = left
        parent.right = right

        heapq.heappush(heap, parent)

    return heap[0]      

text = "AAAAABBC"

root = build_huffman_tree(text)

print("Root frequency:", root.frequency)    
````

````python
def generate_codes(node, current_code="", codes=None):
    if codes is None:
        codes = {}

    # Leaf node
    if node.char is not None:
        codes[node.char] = current_code
        return codes

    # Left branch -> 0
    generate_codes(node.left, current_code + "0", codes)

    # Right branch -> 1
    generate_codes(node.right, current_code + "1", codes)

    return codes
````

````python
codes = generate_codes(root)

print("Huffman codes:")

for char, code in codes.items():
    print(f"{char}: {code}")
````

````python
def encode(text, codes):
    return "".join(codes[char] for char in text)


encoded_text = encode(text, codes)

print("Original text:", text)
print("Encoded text:", encoded_text)
print("Number of encoded bits:", len(encoded_text))
````

````python
def decode(encoded_text, root):
    decoded_text = []
    current_node = root

    for bit in encoded_text:
        if bit == "0":
            current_node = current_node.left
        else:
            current_node = current_node.right

        # We reached a leaf
        if current_node.char is not None:
            decoded_text.append(current_node.char)
            current_node = root

    return "".join(decoded_text)
````

````python
decoded_text = decode(encoded_text, root)

print("Encoded text:", encoded_text)
print("Decoded text:", decoded_text)
print("Original text:", text)
print("Lossless:", decoded_text == text)
````

#### Storing the Huffman Tree

To decompress Huffman-encoded data, the decoder must know the same
Huffman tree (or an equivalent code table) that was used for encoding.

Therefore, the compressed data alone is not always sufficient.
Information describing the tree must also be stored or transmitted.

One approach is to store the frequencies of the symbols. The decoder
can then reconstruct the Huffman tree.

Another approach is to store the structure of the tree or the Huffman
code table directly.

Thus, a compressed file can conceptually contain:

1. Information needed to reconstruct the Huffman tree
2. The encoded binary data

````python
original_bits = len(text) * 8
compressed_bits = len(encoded_text)

compression_percentage = (
    compressed_bits / original_bits
) * 100

space_saved = 100 - compression_percentage

print("Original size:", original_bits, "bits")
print("Compressed size:", compressed_bits, "bits")
print("Compressed / original:", compression_percentage, "%")
print("Space saved:", space_saved, "%")
````

#### Huffman vs. LZ77

Huffman coding and LZ77 are both lossless compression methods,
but they use different ideas.

**Huffman coding**
- Uses the frequency of symbols.
- Frequent symbols receive shorter binary codes.
- Rare symbols receive longer binary codes.
- It is based on probability and information theory.

**LZ77**
- Looks for repeated sequences in previously processed data.
- Repeated sequences can be represented by references instead of
  being stored again.
- It is especially useful when the data contains repeated patterns.

Therefore, the algorithms exploit different kinds of redundancy:
Huffman coding exploits differences in symbol frequencies, while
LZ77 exploits repeated sequences.

In practice, compression formats can combine ideas from both
dictionary-based compression and entropy coding.
